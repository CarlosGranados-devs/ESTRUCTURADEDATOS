## Caso 1: Sistema de Atención a Clientes en un Banco (Patrón FIFO)

**¿Por qué FIFO?**
En un banco, la fila de atención al cliente debe respetar el orden de llegada: el primer cliente que llega es el primero en ser atendido. Este comportamiento es exactamente el que ofrece una **cola (queue)**, por lo que el patrón adecuado es **FIFO**, no LIFO (una pila atendería primero al último que llegó, lo cual sería injusto para un banco).

**Estructura utilizada:** `collections.deque`, porque a diferencia de una lista normal, `deque` permite quitar elementos del inicio (`popleft()`) en tiempo O(1), mientras que `list.pop(0)` es O(n). Esto la hace la estructura ideal para implementar colas de forma eficiente.

**Métodos utilizados:**
- `append()` → para que un cliente **ingrese a la fila** (encolar).
- `popleft()` → para **atender al cliente más antiguo** (desencolar).
- `len()` e indexado (`cola[0]`) → para consultar el estado de la fila sin alterarla.

```python
from collections import deque

class BancoFIFO:
    """
    Simula la fila de atención a clientes de un banco usando el patrón FIFO.
    Se usa deque en lugar de list porque append()/popleft() son O(1),
    ideal para encolar y desencolar clientes de forma eficiente.
    """

    def __init__(self):
        self.fila = deque()          # Cola de clientes en espera
        self.atendidos = []          # Historial de clientes ya atendidos

    def llegar_cliente(self, nombre):
        """Encolar: el cliente llega y se une al final de la fila."""
        self.fila.append(nombre)
        print(f"-> Llega '{nombre}'. Fila actual: {list(self.fila)}")

    def atender_cliente(self):
        """Desencolar: se atiende al cliente que lleva más tiempo esperando."""
        if not self.fila:
            print("No hay clientes en espera.")
            return None

        cliente = self.fila.popleft()   # Se retira el primero en llegar
        self.atendidos.append(cliente)
        print(f"<- Se atiende a '{cliente}'. Fila restante: {list(self.fila)}")
        return cliente

    def clientes_en_espera(self):
        """Consulta sin modificar la fila."""
        return len(self.fila)

    def proximo_en_fila(self):
        """Ve quién sigue sin sacarlo de la fila."""
        return self.fila[0] if self.fila else None


# --- Simulación del caso de uso ---
banco = BancoFIFO()

print("=== Llegada de clientes ===")
banco.llegar_cliente("Carlos")
banco.llegar_cliente("Ana")
banco.llegar_cliente("Luis")
banco.llegar_cliente("Marta")

print("\n=== Atención en ventanilla ===")
banco.atender_cliente()   # Debe atender primero a Carlos (fue el primero en llegar)
banco.atender_cliente()   # Luego a Ana

print(f"\nClientes aún en espera: {banco.clientes_en_espera()}")
print(f"Próximo en ser atendido: {banco.proximo_en_fila()}")
print(f"Historial de atendidos (orden de llegada): {banco.atendidos}")
```

## Caso 2: Verificación de Requisitos Académicos (Álgebra de Sets)

**¿Por qué Sets?**
Para inscribir una materia, un estudiante debe haber aprobado **todas** las materias prerrequisito, sin importar el orden ni si hay repeticiones en sus registros. Un **set** es ideal porque:
- Elimina automáticamente duplicados (si una materia aprobada quedó registrada dos veces, no afecta el resultado).
- Permite operaciones de álgebra de conjuntos (`issubset`, diferencia `-`, intersección `&`) que resuelven directamente el problema de "¿cumple todos los requisitos?" y "¿qué le falta?".

**Métodos utilizados:**
- `issubset()` → para verificar si **todos** los prerrequisitos están cubiertos.
- Diferencia (`-`) → para calcular **qué materias le faltan** al estudiante.
- Intersección (`&`) → para ver qué materias **comparten** dos estudiantes (por ejemplo, para formar grupos de estudio).

```python
def puede_inscribir(prerrequisitos, materias_aprobadas):
    """
    Determina si un estudiante cumple TODOS los prerrequisitos de una materia.
    Se usa issubset(): verifica que 'prerrequisitos' esté completamente
    contenido dentro de 'materias_aprobadas'.
    """
    return prerrequisitos.issubset(materias_aprobadas)


def materias_faltantes(prerrequisitos, materias_aprobadas):
    """
    Calcula qué materias le faltan al estudiante mediante la diferencia
    de conjuntos: prerrequisitos - materias_aprobadas.
    """
    return prerrequisitos - materias_aprobadas


def materias_en_comun(aprobadas_a, aprobadas_b):
    """
    Obtiene las materias que dos estudiantes ya tienen en común,
    útil por ejemplo para armar grupos de estudio.
    """
    return aprobadas_a & aprobadas_b


#Datos del caso de uso
#Prerrequisitos para inscribir Estructuras de Datos
prerrequisitos_ed = {"Programación I", "Programación II", "Matemática Discreta"}

# Materias que cada estudiante ya aprobó
aprobadas_carlos = {"Programación I", "Programación II", "Matemática Discreta", "Física I"}
aprobadas_ana = {"Programación I", "Matemática Discreta"}

print("=== Verificación de inscripción a 'Estructuras de Datos' ===")

print(f"¿Carlos puede inscribir? {puede_inscribir(prerrequisitos_ed, aprobadas_carlos)}")
print(f"¿Ana puede inscribir? {puede_inscribir(prerrequisitos_ed, aprobadas_ana)}")

faltantes_ana = materias_faltantes(prerrequisitos_ed, aprobadas_ana)
print(f"Materias que le faltan a Ana: {faltantes_ana}")

comunes = materias_en_comun(aprobadas_carlos, aprobadas_ana)
print(f"\nMaterias que Carlos y Ana tienen en común: {comunes}")
```
